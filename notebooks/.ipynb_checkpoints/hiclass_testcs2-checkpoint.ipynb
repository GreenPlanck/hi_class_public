```python
cosmo={'omega_b':0.0223828,'omega_cdm':0.1201075,'h':0.67810,'A_s':2.100549e-09,'n_s':0.9660499,'tau_reio':0.05430842,'Omega_Lambda':0,
      'N_ncdm':0,'N_ur':3.04}
```

```python
M=Class()
hiset={
    'Omega_smg':-1,
    'Omega_fld':0,
    'Omega_Lambda':0,
    
    'gravity_model':'propto_omega',
    'parameters_smg':'0.,0.7,0,0,1',
    'expansion_model':'w0wa',
    'expansion_smg':'0.7,-1.3,-0.7',
    #'expansion_model':'lcdm',
    #'expansion_smg':'0.6',
    'skip_stability_tests_smg':'n',
    # 'cs2_safe_smg':10.,
    #'D_safe_smg':10.,
    # 'ct2_safe_smg':10.,
    # 'M2_safe_smg':10.,
    #'method_qs_smg':'automatic'
}
M.set(cosmo)
M.set(hiset)
M.set({'output':'tCl,mPk'})
M.compute()
```

```python
z = M.get_background()['z']
rho_smg=M.get_background()['(.)rho_smg']
rho_crit=M.get_background()['(.)rho_crit']
print(rho_smg[-1]/rho_crit[-1],'\n',M.get_background()['c_s^2'])
```

```python
k = np.logspace(-3,0,100)
z=0.
pk = np.array([M.pk(ki,z) for ki in k])
```

```python
np.savetxt('hiclass_w-0.7.txt',pk)
```

```python
def get_cl_pk(engine,z=0):
    cl = engine.lensed_cl()
    
    spectrum={}
    spectrum.update(cl)
    spectrum['k']=k
    spectrum['pk']=pk
    return spectrum
get_cl_pk(M)
```
